```python
import matplotlib.pyplot as plt

# Data from the graph
years = [2021, 2022, 2023, 2024]
market_revenue = [58.4, 59.2, 59.8, 61.1]

# Create the graph
plt.figure(figsize=(16, 9))

plt.plot(
    years,
    market_revenue,
    marker="o",
    linewidth=2.5,
    markersize=8
)

# Title
plt.title(
    "Germany Telecommunications Market Revenue",
    fontsize=20
)

# Axis labels
plt.xlabel(
    "Year",
    fontsize=14
)

plt.ylabel(
    "Revenue (€bn)",
    fontsize=14
)

# Grid
plt.grid(
    True,
    alpha=0.5
)

# Make the x-axis show the years as whole numbers
plt.xticks(years)

# Improve layout
plt.tight_layout()

# Display
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data from the graph
years = ["FY2021", "FY2022", "FY2023", "FY2024", "FY2025"]

total_revenue = [43.8, 37.0, 37.7, 36.7, 37.4]
service_revenue = [39.2, 30.2, 30.3, 29.9, 30.8]

# Create figure
plt.figure(figsize=(16, 9))

# Plot total revenue
plt.plot(
    years,
    total_revenue,
    marker="o",
    linewidth=2.5,
    markersize=8,
    label="Total Revenue"
)

# Plot service revenue
plt.plot(
    years,
    service_revenue,
    marker="o",
    linewidth=2.5,
    markersize=8,
    label="Service Revenue"
)

# Title and axis labels
plt.title(
    "Vodafone Group Revenue Growth",
    fontsize=20
)

plt.xlabel(
    "Financial Year",
    fontsize=14
)

plt.ylabel(
    "Revenue (€bn)",
    fontsize=14
)

# Grid
plt.grid(
    True,
    alpha=0.5
)

# Legend
plt.legend(
    fontsize=13,
    loc="upper right"
)

# Improve layout
plt.tight_layout()

# Display
plt.show()
```
